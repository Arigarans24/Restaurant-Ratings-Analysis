```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("/content/drive/MyDrive/Dataset .csv")

print("First 5 Records:")
print(data.head())

print("\nAnalyzing Aggregate Ratings...")

plt.figure(figsize=(8, 5))
plt.hist(data["Aggregate rating"], bins=10, edgecolor="black")
plt.title("Distribution of Aggregate Ratings")
plt.xlabel("Aggregate Rating")
plt.ylabel("Number of Restaurants")
plt.show()

rating_groups = pd.cut(
    data["Aggregate rating"],
    bins=[0, 1, 2, 3, 4, 5],
    labels=["0-1", "1-2", "2-3", "3-4", "4-5"],
    include_lowest=True
)

rating_count = rating_groups.value_counts().sort_index()

print("\nNumber of Restaurants in Each Rating Range:")
print(rating_count)

common_range = rating_count.idxmax()
common_count = rating_count.max()

print("\nMost Common Rating Range:", common_range)
print("Restaurants in this Range:", common_count)

avg_votes = data["Votes"].mean()

print("\nAverage Number of Votes Received by Restaurants:")
print(round(avg_votes, 2))

print("\n----- Summary -----")
print(f"Most Common Rating Range : {common_range}")
print(f"Average Votes Received   : {avg_votes:.2f}")
```
